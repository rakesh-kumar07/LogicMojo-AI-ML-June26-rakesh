# Bank Loan Risk — EDA Case Study

## **1. Introduction**

**Business understanding:** Consumer finance companies often struggle to assess
applicants with thin or no credit history. Some of these applicants turn out to
be reliable payers; others become defaulters. Two risks sit on either side of
every approval decision:

- **Type 1 risk** — rejecting an applicant who would have repaid (lost business).
- **Type 2 risk** — approving an applicant who defaults (financial loss).

Every application in the dataset resolves into one of two outcomes:
- **Payment difficulty** — late payment beyond a threshold on early instalments.
- **All other cases** — instalments paid on time.

And every application passes through one of four decision states: **Approved,
Cancelled, Refused, Unused offer**.

**Business objective:** identify the applicant and loan attributes that most
strongly drive default, so the business can approve, reject, resize, or reprice
loans more accurately — without turning away good-faith borrowers.

**The Core Pattern: Identical Missingness**
- Every single one of these columns has exactly 41,519 missing values. This identical count strongly suggests that when Credit Bureau data is missing, it is missing across the board for that applicant. This usually means the applicant simply has no credit history or record with the bureau.

**Why Mean or Median Fails Here**
- The Median is 0 for almost all columns (except _YEAR which is 1). Imputing the median would just fill everything with zeros anyway.
- The Mean is heavily skewed by rare extreme outliers (e.g., someone with 261 inquiries in a quarter!). Using the mean would fill missing records with fractional values like 0.267, which makes no sense for discrete inquiry counts.

**Impute with 0 (Most common)** 
- If an applicant has no recorded inquiries, it is highly likely they had zero inquiries. Filling these with 0 treats them as low-risk, clean-slate applicants.

```python
cb_cols = [
    "AMT_REQ_CREDIT_BUREAU_HOUR", "AMT_REQ_CREDIT_BUREAU_DAY", 
    "AMT_REQ_CREDIT_BUREAU_WEEK", "AMT_REQ_CREDIT_BUREAU_MON", 
    "AMT_REQ_CREDIT_BUREAU_QRT", "AMT_REQ_CREDIT_BUREAU_YEAR"
]

# Fill missing credit bureau inquiries with 0
application_df[cb_cols] = application_df[cb_cols].fillna(0)

print(application_df[["AMT_REQ_CREDIT_BUREAU_HOUR","AMT_REQ_CREDIT_BUREAU_DAY","AMT_REQ_CREDIT_BUREAU_WEEK","AMT_REQ_CREDIT_BUREAU_MON","AMT_REQ_CREDIT_BUREAU_QRT","AMT_REQ_CREDIT_BUREAU_YEAR"]].isna().sum())
```

```python
print(application_df[["OBS_30_CNT_SOCIAL_CIRCLE", "DEF_30_CNT_SOCIAL_CIRCLE", "OBS_60_CNT_SOCIAL_CIRCLE", "DEF_60_CNT_SOCIAL_CIRCLE"]].isna().sum())
application_df[["OBS_30_CNT_SOCIAL_CIRCLE", "DEF_30_CNT_SOCIAL_CIRCLE", "OBS_60_CNT_SOCIAL_CIRCLE", "DEF_60_CNT_SOCIAL_CIRCLE"]].describe()
```

1. The Core Pattern: Extreme Zero-Inflation
- The Median (50%) is 0 across all four columns.
- For the default columns (DEF_30 and DEF_60), even the 75th percentile is 0.This indicates that the vast majority of applicants have zero observed observations or defaults in their social circle. 
- Imputing with 0 accurately captures the most typical, representative state of the data.

2. Why Mean Imputation Fails
- The means are heavily dragged upward by massive, rare outliers (e.g., a maximum of 348 observations or 34 defaults).
- Imputing with the mean would fill the missing entries with fractions like 1.42 or 0.14. Because these are discrete count variables (you cannot have 0.14 of a person defaulting), fractional imputation makes little structural sense.

3. Missingness Percentage is Tiny
- There are exactly 1,021 missing rows out of over 307,000 records.This represents roughly 0.33% of your data. Because it is so small, filling it with the most common value (0) preserves the underlying distribution without introducing bias.

1. **NAME_TYPE_SUITE (1,292 missing values ~ 0.4%)**
- Good Choice: Mode Imputation (Impute with 'Unaccompanied').
- Why: The missingness is extremely low (0.4%). Furthermore, 'Unaccompanied' is overwhelmingly dominant, making up over 81% of the known data (freq: 248,526 out of count: 306,219). Imputing with the mode is highly safe here.
2. **OCCUPATION_TYPE (96,391 missing values ~ 31.3%)**
- Good Choice: Constant Imputation (Impute with a new category: 'Unknown').
- Why: This column is missing nearly one-third of its data. Forcing the mode ('Laborers') would massive bias your data. In credit risk modeling, a missing occupation usually means the applicant is unemployed, retired, a student, or works in an unlisted sector. Treating 'Unknown' as its own distinct category allows your machine learning model to learn the specific risk profile of this group.
3. **YEARS_EMPLOYED (83,278 missing values ~ 27.1%)**
- Good Choice: Constant Imputation (Impute with a new category: 'Missing' or 'Unknown').
- Why: Like occupation, this has massive missingness (27%). Imputing the most frequent bin ('0-5') would introduce severe artificial bias. Keeping it as a separate category preserves the structural reality that over a quarter of your applicants do not have a recorded employment duration (often correlating directly with the missing occupation rows).

```python
# 1. Mode Imputation for the low-missing column
mode_suite = application_df['NAME_TYPE_SUITE'].mode()[0]
application_df['NAME_TYPE_SUITE'] = application_df['NAME_TYPE_SUITE'].fillna(mode_suite)

# 2. Handle OCCUPATION_TYPE
if application_df['OCCUPATION_TYPE'].dtype.name == 'category':
    application_df['OCCUPATION_TYPE'] = application_df['OCCUPATION_TYPE'].cat.add_categories('Unknown')
application_df['OCCUPATION_TYPE'] = application_df['OCCUPATION_TYPE'].fillna('Unknown')

# 3. Handle YEARS_EMPLOYED
if application_df['YEARS_EMPLOYED'].dtype.name == 'category':
    application_df['YEARS_EMPLOYED'] = application_df['YEARS_EMPLOYED'].cat.add_categories('Unknown')
application_df['YEARS_EMPLOYED'] = application_df['YEARS_EMPLOYED'].fillna('Unknown')

```

1. **AMT_ANNUITY and AMT_GOODS_PRICE (Continuous Numerical)**
- Good Choice: Median Imputation + a Missingness Indicator Column.

However, because 22% of your data is missing due to application status, adding a tracking column prevents the model from treating these missing rows like normal average loans.
2. **CNT_PAYMENT (Discrete/Integer Count)**
- Good Choice: Constant Imputation with 0.
- Why: This column tracks the term length of the loan in months. If an application was canceled or refused at the start, the true number of payments made or scheduled is literally 0. Imputing a fraction (like the mean 16.05) makes no structural sense for a discrete month count.

```python
prev_cols = ['AMT_ANNUITY', 'AMT_GOODS_PRICE', 'CNT_PAYMENT']

# 1. Create flags to let your ML model learn WHY the data was missing
for col in prev_cols:
    previous_df[f'{col}_IS_NULL'] = previous_df[col].isna().astype(int)

# 2. Impute continuous columns with their respective medians
previous_df['AMT_ANNUITY'] = previous_df['AMT_ANNUITY'].fillna(previous_df['AMT_ANNUITY'].median())
previous_df['AMT_GOODS_PRICE'] = previous_df['AMT_GOODS_PRICE'].fillna(previous_df['AMT_GOODS_PRICE'].median())

# 3. Impute the term count with 0
previous_df['CNT_PAYMENT'] = previous_df['CNT_PAYMENT'].fillna(0)

print(previous_df[["AMT_ANNUITY","AMT_GOODS_PRICE","CNT_PAYMENT"]].isna().sum())
```

```python
def plot_categorical_segmented(feature, ylog=False, horizontal_layout=True):
    cat_perc = (application_df[[feature, 'TARGET']]
                .groupby(feature, as_index=False)['TARGET']
                .mean())
    cat_perc['TARGET'] *= 100
    cat_perc.sort_values(by='TARGET', ascending=False, inplace=True)

    figsize = (18, 6) if horizontal_layout else (24, 30)
    fig, (ax1, ax2) = plt.subplots(
        ncols=2 if horizontal_layout else 1,
        nrows=1 if horizontal_layout else 2,
        figsize=figsize
    )

    # --- Subplot 1: Count plot ---
    sns.countplot(ax=ax1, x=feature, data=application_df, hue='TARGET',
                   order=cat_perc[feature], palette=['g', 'r'])
    ax1.set_title(feature, fontdict={'fontsize': 10, 'fontweight': 3, 'color': 'Blue'})
    ax1.legend(['Repayer', 'Defaulter'])
    ax1.set_xticklabels(ax1.get_xticklabels(), rotation=90)

    if ylog:
        ax1.set_yscale('log')
        ax1.set_ylabel("Count (log)", fontdict={'fontsize': 10, 'fontweight': 3, 'color': 'Blue'})

    # --- Subplot 2: Defaulter percentage ---
    sns.barplot(ax=ax2, x=feature, y='TARGET', order=cat_perc[feature],
                data=cat_perc, palette='Set2')
    ax2.set_ylabel('Percent of Defaulters [%]', fontsize=10)
    ax2.set_title(f"{feature} Defaulter %", fontdict={'fontsize': 15, 'fontweight': 5, 'color': 'Blue'})
    ax2.tick_params(axis='both', which='major', labelsize=10)
    ax2.set_xticklabels(ax2.get_xticklabels(), rotation=90)

    plt.tight_layout()
    plt.show()
```
